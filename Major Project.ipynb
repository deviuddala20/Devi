```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("Retail_Sales_Cleaned.xlsx")
```

```python
df.head()
```

```python
df.info()
```

```python
df["OrderDate"] = pd.to_datetime(df["OrderDate"])
```

```python
df["Year"] = df["OrderDate"].dt.year
df["Month"] = df["OrderDate"].dt.strftime("%b")
df["MonthNo"] = df["OrderDate"].dt.month
df["Quarter"] = df["OrderDate"].dt.quarter
```

```python
month_order = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']

df["Month"] = pd.Categorical(
    df["Month"],
    categories=month_order,
    ordered=True
)
```

```python
total_revenue = df["Revenue"].sum()
total_orders = df["OrderID"].count()
total_quantity = df["Quantity"].sum()
gross_sales = df["GrossAmount"].sum()
avg_order = total_revenue / total_orders

print("Total Revenue :", total_revenue)
print("Total Orders :", total_orders)
print("Total Quantity :", total_quantity)
print("Gross Sales :", gross_sales)
print("Average Order :", avg_order)
```

```python
monthly = df.groupby("Month")["Revenue"].sum()

plt.figure(figsize=(10,5))
plt.plot(monthly.index, monthly.values, marker='o')
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)

plt.savefig("Monthly_Revenue_Trend.png")
plt.show()
```

```python
category = df.groupby("Category")["Revenue"].sum()

plt.figure(figsize=(8,5))
plt.bar(category.index, category.values)

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("Revenue_by_Category.png")
plt.show()
```

```python
region = df.groupby("Region")["Revenue"].sum()

plt.figure(figsize=(7,7))

plt.pie(
    region.values,
    labels=region.index,
    autopct='%1.1f%%'
)

plt.title("Revenue by Region")

plt.savefig("Revenue_by_Region.png")

plt.show()
```

```python
top_products = (
    df.groupby("Product")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))

plt.barh(top_products.index, top_products.values)

plt.title("Top 10 Products")

plt.xlabel("Revenue")

plt.gca().invert_yaxis()

plt.tight_layout()

plt.savefig("Top10_Products.png")

plt.show()
```

```python
payment = df.groupby("PaymentMethod")["Revenue"].sum()

plt.figure(figsize=(7,7))

plt.pie(
    payment.values,
    labels=payment.index,
    autopct='%1.1f%%'
)

plt.title("Revenue by Payment Method")

plt.savefig("Payment_Method.png")

plt.show()
```

```python
salesrep = (
    df.groupby("SalesRep")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))

plt.bar(salesrep.index, salesrep.values)

plt.title("Sales Representative Performance")

plt.xlabel("Sales Representative")

plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("SalesRep_Performance.png")

plt.show()
```

```python
qty = df.groupby("Category")["Quantity"].sum()

plt.figure(figsize=(8,5))

plt.bar(qty.index, qty.values)

plt.title("Quantity Sold by Category")

plt.xlabel("Category")

plt.ylabel("Quantity")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("Quantity_by_Category.png")

plt.show()
```

```python
quarter = df.groupby("Quarter")["Revenue"].sum()

plt.figure(figsize=(7,5))

plt.bar(
    ["Q1","Q2","Q3","Q4"],
    quarter.values
)

plt.title("Quarterly Revenue")

plt.xlabel("Quarter")

plt.ylabel("Revenue")

plt.savefig("Quarterly_Revenue.png")

plt.show()
```

```python
year = df.groupby("Year")["Revenue"].sum()

plt.figure(figsize=(6,5))

plt.bar(year.index.astype(str), year.values)

plt.title("Yearly Revenue")

plt.xlabel("Year")

plt.ylabel("Revenue")

plt.savefig("Yearly_Revenue.png")

plt.show()
```

```python
orders = df.groupby("Region")["OrderID"].count()

plt.figure(figsize=(8,5))

plt.bar(orders.index, orders.values)

plt.title("Orders by Region")

plt.xlabel("Region")

plt.ylabel("Orders")

plt.savefig("Orders_by_Region.png")

plt.show()
```
